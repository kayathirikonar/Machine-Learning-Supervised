#Important Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn
```

# Read Data

```python
df = pd.read_csv('HR-Employee-Attrition.csv')
```

```python
df
```

```python
df.columns
```

```python
# Print column names to verify their existence and exact spelling
print(df.columns)

# Attempt to drop the columns. Adjust the list if the printed output shows different names.
df = df.drop(['EmployeeCount', 'EmployeeNumber', 'Over18', 'StandardHours','HourlyRate','MonthlyRate'], axis=1)
```

```python
df.columns
```

```python
df.info()
```

```python
df['Attrition'] = df['Attrition'].map({'Yes': 1, 'No': 0})
```

```python
df = pd.get_dummies(df, drop_first=True)
```

```python
df.info()
```

```python
df.head(1)
```

# Decide Dependant and Independant Elements in Data Set

```python
x=df.drop('Attrition',axis=1)
```

```python
x
```

```python
y=df['Attrition']
```

```python
y
```

#Train Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test =train_test_split(x,y,test_size=0.3)
```

```python
X_train.info()
```

```python
X_test.info()
```

#Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
logmodel = LogisticRegression()
```

```python
logmodel.fit(X_train,y_train)
```

```python
predictions = logmodel.predict(X_test)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,predictions))
```

```python
from sklearn.metrics import confusion_matrix
```

```python
print(confusion_matrix(y_test,predictions))
```
